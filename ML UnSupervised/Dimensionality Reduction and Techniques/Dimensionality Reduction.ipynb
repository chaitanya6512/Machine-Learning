### what is dimensionality reduction?
- When working with machine learning models, datasets with too many features can cause issues like slow computation and overfitting.
- It helps us reduce the number of features while training key information.
- It converts high-dimensional data into low dimensional data while preserving important details.
### Dimensional Reduction Technquies:
- It can be broadly divided into two categories.
### 1. Feature Selection:
- Feature selection chooses the most relevant features from the dataset without altering them.
- It helps remove redundant or irrelevant features, improving model performance.
- Some methods are -
#### 1. Filter Methods:
- rank the features based on their relevance to the target variable.
#### 2. Wrapper Methods:
- use the model performance as the criteria for selecting features.
#### 3. Embedded Methods:
- combine feature selection with the model traning process.
### 2. Feature Extraction:
- Feature extraction involves creating new feature by combining or transforming the original features.
- These new features retain most of the dataset's important information in fewer dimensions.
- Common feature extraction methods are -
#### 1.Prinicipal Component Analysis:
- Converts the correlated variables into uncorrelated prinicipal components hence reduceing dimensionality while modeling as much variance as possible enabling more efficient analysis.
#### 2. Independent Component Analysis:
- Identifies statistically independent componets, ideal for application like "blind source seperation" where traditional correlation-based methods fall short. 
#### 3. Missing Value Ratio:
- Missing with variables beyond a set threshold are removed, improving dataset reliability.
#### 4. Forward Feature Selection:
- Starts with all features and removes most significans ones in each iteration.
- The process continues until only the most impactful features remain, optimizing model performance.
#### 5. Backward Feature Elimination:
- It start with one feature, adds others incrementally and keeps those improving model performance.
#### 6. Random Forest:
- Random forest uses decision tress to evaluate feature importance, automatically selecting the most relevant features without need for manual coding, enhancing model accuracy.
#### 7. Factor Analysis:
- Groups variables by coorelation and keeps the most relevant ones for further analysis.
### Real-World Uses:
- Text Categorization
- Intrusion Detection
- Gene Expression Analysis
- Image Retrieval
### Advantages:
- Faster Computation
- Better Visualization
- Prevent Overfitting
### Disadvantages:
- Data Loss and Reduced Accuracy
- Choosing Right Components.
