### Association Rule:
- Association rules are fundamental concepts used to find the relationships, correlations or patterns within large sets of data items.
- They describe how often itemsets occur together in transactions and express implications of the form:
  - X → Y.
  - where X and Y are disjoint sets of items.
  - This rule suggests that when items in X appear, items in Y tend to appear as well.
- Association rules originated from market basket analysis and helps retailers and analysts understand customer behavior by discovering item associations in transaction data.
### Key Components:
1. Antecedent (X): The "if" part representing one or more items found in transactions.
2. Consequent (Y): The "then" part representing the items likely to be purchased when antecedent items appear.
### Rule Evaluation Metrics:
#### 1. Support:
- Fraction of transactions containing the itemsets in both X and Y.
- Support(X → Y) = No of transactions with (X∪Y) / Total number of transactions.
- Support measures how frequently the combination appears in the data.
#### 2. Confidence:
- Probability that transactions with X also include Y.
- Confidence (X → Y ) = Support(X∪Y) / Support(X).
- Confidence measures the reliability of the inference.
#### 3. Lift:
- The ratio of observed support to that expected if x and Y were independent.
- Lift(X → Y ) = Confidence(X → Y ) / Support(Y).
  - Here Lift>1 implies a positive association - items occur together morethan expected.
  - Lift=1 implies independence.
  - Lift <1 implies a negative association.

### NetworkX Library:

- It models real-world systems as groups, where nodes represent entities and edges represent relationships.
- The library supports loading, storing and generating various types of graphs, analyzing network structures, designing algorithms and drawing networks with ease and flexibility.

```python
import networkx as nx

# Create an empty graph
G = nx.Graph()
```

### Adding Nodes:
- You can add nodes one at a time or in batches.
  - add_node(1) - Only one node can be added.
  - add_nodes_from([2,3,..]) - Add morethan one nodes.

```python
# Add individual nodes
G.add_node(1)

# Add Multiple_at_a_time
G.add_nodes_from([2,3,4,7,9])
```

### Adding Edges:
- Edges connect the nodes in your graph.
  - add_edge(1,2) - By using this, We can add edge to 1 to 2.
  - add_edges_from([(1,4),...])- By using this we can add mutliple edges at a time.

```python
# Add individual edges
G.add_edge(1, 2)
G.add_edge(3, 1)
G.add_edge(2, 4)
G.add_edge(4, 1)
G.add_edge(9, 1)

# Add multiple edges at once
G.add_edges_from([(1, 7), (2, 9)])
```

### Removing Nodes and Edges:
- You can remove nodes and edges just an easily.
  - G.remove_edge(1,2) - By using this, we can remove edge between 1 and 2.
  - G.remove_node(3) - By using this, we can remove the node 3.

```python
# Removes node 3
G.remove_node(3)

# Removes edge between node 1 and 2.
G.remove_edge(1,2)
```

### Network Analysis with NetworkX:
- NetworkX allows you to perform deep analysis of your grapgh.
  - G.nodes() - returns a list of all nodes currently in the graph.
  - G.edges() - returns a list of all edges (i.e node pairs that are connected).
  - G.number_of_nodes() - returns the count of total nodes.
  - G.number_of_edges() - returns the count of total connections.
  - G.degress(2) - Gives the number of connections the node 2 has (its degree).
  - list(G.neighbors(2)) - lists all nodes that are directly connected to node 2.

```python
# All nodes
print(G.nodes())

# All edges
print(G.edges())

# No of nodes and edges
print(G.number_of_edges())
print(G.number_of_nodes())

# Degree of a node
print(G.degree(2))

# Neighbors of a node
print(list(G.neighbors(2)))

# Clear the graph (remove all nodes and edges)
G.clear()
```
