### Gradient Boosting:
- Gradient Boosting is a boosting algorithm.
- Here each new model is trained to minimize the loss function such as mean squarred error or cross-entropy of the previous model using gradient descent.
- In each iteration algorithm computes the loss function with respects to the predictions and then trains a new weak models to predict the gradient.
- Predictions of the new model are then added to the ensemble(all models prediction) and the process is repeated until stopping criteria is met.
### Shrinkage and Model Complexity:
- A key feature of Gradient Boosting is shrinkage which scales the contribution of each new model using learning rate (denoted as η).
  1. smaller learning rates - The contribution of each tree is smaller which reduces the risk of overfitting but requires more trees to achieve the same performance.
  2. larger learning rates - each tree has more significant impact but this can lead to overfitting.

### Working of Gradient Boosting:
#### 1. Sequential Learning Process:
- The ensemble consists of multiple trees each trained to correct the erros of the previous one.
- In the first iteration Tree 1 is trained on the original data x and the true labels y.
- It makes predictions which are used to compute the errors.
#### 2. Residuals Calculations:
- In the second iteration Tree 2 is trained using the feature matrix x and the erros from Tree 1 as labels.
- This means Tree 2 ia trained to predict the erros of the Tree 1.
- This process continues for all the trees in the ensemble.
- Each subsequent tree is trained to predict the errors of the previous tree.
#### 3. Shrinkage:
- After each tree is trained its predictions are shrunk by multiplying them with the learning rate η which ranges from 0 to 1.
- This prevents overfitting by ensuring each tree has a smaller impact on the final impact.
- Once all trees are obtained predictions are made by summing the contributions of all the trees.
  - y pred = y1 + η.r1 + η.r2 +  + η.rN

### Implementaton of Gradient Boosting using Classification:
- We use Gradient Boosting Classifier to predict digits from Digits dataset.
  1. Import necessary libraries
  2. Setting SEED for producibility.
  3. Load the digit dataset and split it into train and test.
  4. Instantiate Gradient Boosting classifier and fit the model.
  5. Predict the test set and compute the accuracy score.

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score

SEED = 23

X, y = load_digits(return_X_y = True)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)

gradient_boosting = GradientBoostingClassifier(n_estimators=300, 
                                               learning_rate=0.05, 
                                               random_state=100,
                                               max_features=5)
gradient_boosting.fit(X_train, y_train)

y_pred = gradient_boosting.predict(X_test)

acc= accuracy_score(y_test, y_pred)
print("Gradient Boosting Classifier accuracy is:{:2f}".format(acc))
```

### Implementation of Gradient Boosting using Regression:
- We use Gradient Boosting Regressor on the Diabetes dataset to predict continous values.
  1. Import the necessary libraries.
  2. Seeting the SEED for reproducability.
  3. Load the diabetes dataset and split it into train and test.
  4. Instantiate Gradient Boosting Regressor and fit model.
  5. Predict on the test set and compute RMSE.

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error

SEED = 23

X, y = load_diabetes(return_X_y = True)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

gbr = GradientBoostingRegressor(loss='absolute_error',
                               learning_rate=0.1,
                               n_estimators=300,
                               max_depth=1,
                               random_state=SEED,
                               max_features=5)
gbr.fit(X_train, y_train)
y_pred = gbr.predict(X_test)

test_mse = mean_squared_error(y_test, y_pred) ** (1/2)

print("Root mean sqaured error:{:.2f}".format(test_mse))
```
